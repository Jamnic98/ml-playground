```python
import numpy as np
from typing import Tuple

# values 1 -> 9
valid_board_square_values = range(1, 10)

class Board:
    def __init__(self, size: int):
        self.size = None
        self.squares = None
        try:
            if size not in [9]:
                raise ValueError(f"Invalid board size: {size}")
            self.squares = np.zeros((size, size), dtype='int')
            self.size = size
        except ValueError as e:
            print(e)
            return None

    def update(self, coords: Tuple[int, int], value: int) -> None:
        board_size_range = range(1, self.size + 1)
        try:
            x, y = coords
            if x not in board_size_range or y not in board_size_range:
                raise ValueError(f"Invalid coords: {coords}")
            if value not in valid_board_square_values:
                raise ValueError(f"Invalid value: {value}")
            self.squares[y - 1][x - 1] = value
        except ValueError as e:
            print(e)

    def display(self):
        """Display the Sudoku grid with dividers."""
        horizontal_divider = "+-------+-------+-------+"
        print(horizontal_divider)  # Top divider

        for i, row in enumerate(self.squares[::-1]):
            row_display = "| " + " | ".join(
                " ".join(str(num) if num != 0 else "_" for num in row[start:start+3])
                for start in range(0, self.size, 3)
            ) + " |"
            print(row_display)

            # Add horizontal dividers every 3 rows
            if (self.size - i - 1) % 3 == 0:
                print(horizontal_divider)

# Example usage
board = Board(9)
board.update((1, 3), 3)
board.update((4, 3), 3)
board.display()
```
